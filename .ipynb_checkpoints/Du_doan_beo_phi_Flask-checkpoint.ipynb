# Dự đoán mức béo phì bằng mạng nơ-ron

Cấu trúc giống bài tiểu đường: **đọc dữ liệu → chia train/test → chuẩn hoá → xây mạng nơ-ron → huấn luyện → đánh giá → web Flask**.

Khác với bài tiểu đường (2 lớp, 1 nơ-ron `sigmoid`), bài này có **7 mức cân nặng** nên:
- Lớp ra có **7 nơ-ron `softmax`** (thay vì 1 nơ-ron `sigmoid`).
- Hàm mất mát là `sparse_categorical_crossentropy` (thay vì `binary_crossentropy`).
- Dữ liệu có **cột chữ** (Gender, CAEC, MTRANS...) nên phải mã hoá thành số trước khi chuẩn hoá.

> **Lưu ý quan trọng:** nhãn được tính từ BMI = cân nặng / chiều cao². Nếu đưa `Height` và `Weight` vào mô hình thì nó chỉ "tính lại BMI". Vì vậy biến `DUNG_CHIEU_CAO_CAN_NANG` ở cell 2 mặc định là `False` (dự đoán từ thói quen). Đổi thành `True` nếu muốn dùng thêm hai cột đó.

## 1.1. Đọc dữ liệu

```python
import pandas as pd
import numpy as np

path = 'obesity_level.csv'
# path = '/content/drive/MyDrive/Colab Notebooks/obesity_level.csv'   # nếu file để trên Google Drive
df = pd.read_csv(path)
print(df.shape)
df.head()
```

## 1.2. Làm sạch, tách X và y bằng `iloc`, chia train (85%) và test (15%)

File `obesity_level.csv` có vài điểm cần xử lý:
- Cột đầu tiên là `id` (số thứ tự, không mang thông tin) nên bỏ đi.
- **Cột cuối là nhãn** (tên cột bị lỗi thành `0be1dad`), vì vậy dùng `iloc` để lấy theo vị trí, không cần gõ tên cột.
- Trong file, chữ `no` ở hai cột `CAEC`, `CALC` bị thay bằng `0`, và nhãn `Normal_Weight` bị ghi thành `0rmal_Weight`. Code bên dưới sửa lại cho đúng.
- Nhãn là chữ nên đổi sang số 0 đến 6 theo thứ tự tăng dần của cân nặng.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import joblib

DUNG_CHIEU_CAO_CAN_NANG = False     # False: chỉ dùng thói quen; True: dùng cả Height, Weight (gần như tính lại BMI)

ORDER = ['Insufficient_Weight', 'Normal_Weight', 'Overweight_Level_I', 'Overweight_Level_II',
         'Obesity_Type_I', 'Obesity_Type_II', 'Obesity_Type_III']

# Sửa lỗi chữ của file: "no" bị thay thành "0", "Normal_Weight" bị ghi thành "0rmal_Weight"
for c in ['CAEC', 'CALC']:
    df[c] = df[c].replace('0', 'no')
df.iloc[:, -1] = df.iloc[:, -1].replace('0rmal_Weight', 'Normal_Weight')

# Các mức độ vốn là số nguyên, làm tròn lại
for c in ['FCVC', 'NCP', 'CH2O', 'FAF', 'TUE']:
    df[c] = df[c].round()

X = df.iloc[:, 1:-1]       # bỏ cột id (cột đầu) và cột nhãn (cột cuối)
y = df.iloc[:, -1]         # cột cuối là nhãn
y = y.map({n: i for i, n in enumerate(ORDER)}).astype(int)
print("Nhãn:", y.value_counts().sort_index().to_dict())

if not DUNG_CHIEU_CAO_CAN_NANG:
    X = X.drop(columns=['Height', 'Weight'])

X = pd.get_dummies(X, dtype=float)                   # mã hoá cột chữ thành cột 0/1
feature_names = list(X.columns)
print("Số đặc trưng sau mã hoá:", len(feature_names))

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=42, stratify=y)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

joblib.dump(scaler, 'scaler.pkl')
joblib.dump(feature_names, 'feature_names.pkl')
print("Đã lưu scaler.pkl và feature_names.pkl")
```

## 2. Xây dựng mạng nơ-ron

Mạng gồm: Input, **2 lớp ẩn** (64 và 32 nơ-ron, hàm kích hoạt `relu`), và lớp ra **7 nơ-ron `softmax`** (mỗi nơ-ron là xác suất của một mức cân nặng).

```python
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.models import Sequential

n_features = X_train.shape[1]
n_classes = 7

model = Sequential([
    Input(shape=(n_features,)),
    Dense(64, activation="relu"),
    Dense(32, activation="relu"),
    Dense(n_classes, activation="softmax"),
])

model.compile(loss="sparse_categorical_crossentropy",   # nhãn là số nguyên 0..6
              optimizer="adam",
              metrics=["accuracy"])
model.summary()
```

## 3. Huấn luyện

```python
from tensorflow import keras

# Dừng sớm khi val_loss không cải thiện trong 10 epoch, và lấy lại trọng số tốt nhất
early_stop = keras.callbacks.EarlyStopping(monitor="val_loss", patience=10, restore_best_weights=True)

history = model.fit(X_train, y_train,
                    epochs=100,            # tối đa 100 vòng lặp qua toàn bộ tập train
                    batch_size=32,         # số mẫu xử lý trước khi cập nhật trọng số 1 lần
                    validation_split=0.2,  # 20% tập train dùng để theo dõi validation
                    callbacks=[early_stop],
                    verbose=1)

model.save('modelNN.h5')
print("Đã lưu model vào file modelNN.h5")
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(history.history['loss'], label='train'); ax[0].plot(history.history['val_loss'], '--', label='val')
ax[1].plot(history.history['accuracy'], label='train'); ax[1].plot(history.history['val_accuracy'], '--', label='val')
ax[0].set_title('Loss theo epoch'); ax[1].set_title('Accuracy theo epoch')
for a in ax: a.set_xlabel('Epoch'); a.legend()
plt.tight_layout(); plt.show()
```

## 4. Đánh giá trên tập test

```python
from tensorflow.keras.models import load_model
from sklearn.metrics import classification_report, ConfusionMatrixDisplay

loaded_model = load_model('modelNN.h5', compile=False)

pred = loaded_model.predict(X_test, verbose=0).argmax(axis=1)
print(f"Test Accuracy: {(pred == y_test.values).mean():.4f}")
print(classification_report(y_test, pred, target_names=ORDER, digits=3))
```

```python
fig, ax = plt.subplots(figsize=(8, 7))
ConfusionMatrixDisplay.from_predictions(y_test, pred, display_labels=ORDER, normalize='true',
                                        xticks_rotation=45, cmap='Blues', values_format='.2f', ax=ax, colorbar=False)
ax.set_title('Ma trận nhầm lẫn (tỉ lệ theo lớp thật)')
plt.tight_layout(); plt.show()
```

## 5. Hàm dự đoán cho một người mới

Người dùng nhập một từ điển thông tin. Hàm mã hoá cột chữ giống lúc huấn luyện rồi `reindex` theo `feature_names` (cột nào thiếu thì điền 0), sau đó chuẩn hoá và dự đoán.

```python
scaler = joblib.load('scaler.pkl')
feature_names = joblib.load('feature_names.pkl')
model = loaded_model

NHI_PHAN = ['family_history_with_overweight', 'FAVC', 'SMOKE', 'SCC']   # trong file đã là 0/1

def du_doan(nguoi):
    nguoi = dict(nguoi)
    for c in NHI_PHAN:
        if isinstance(nguoi.get(c), str):                    # form gửi 'yes'/'no' thì đổi sang 1/0
            nguoi[c] = 1 if nguoi[c] == 'yes' else 0
    x = pd.get_dummies(pd.DataFrame([nguoi]), dtype=float)
    x = x.reindex(columns=feature_names, fill_value=0.0)      # đúng thứ tự và đủ cột như lúc huấn luyện
    proba = model.predict(scaler.transform(x), verbose=0)[0]
    return ORDER[int(proba.argmax())], {n: round(float(p) * 100, 1) for n, p in zip(ORDER, proba)}

nguoi_moi = {'Gender': 'Male', 'Age': 28, 'Height': 1.75, 'Weight': 80,
             'family_history_with_overweight': 'yes', 'FAVC': 'yes', 'FCVC': 2, 'NCP': 3,
             'CAEC': 'Frequently', 'SMOKE': 'no', 'CH2O': 1, 'SCC': 'no', 'FAF': 0, 'TUE': 2,
             'CALC': 'Sometimes', 'MTRANS': 'Automobile'}
nhan, xs = du_doan(nguoi_moi)
print("Dự đoán:", nhan)
xs
```

## 6. Web demo bằng Flask (+ ngrok để mở trên Colab)

Flask phục vụ một trang HTML và một đường dẫn `/predict` nhận dữ liệu JSON.

**Trước khi chạy:** lấy token miễn phí tại ngrok.com rồi dán vào `NGROK_TOKEN`. **Không chia sẻ notebook có chứa token thật** (ai có token đều dùng được tài khoản ngrok của bạn); nếu lỡ lộ thì đổi token mới trong trang quản lý ngrok.

```python
%pip install -q flask flask-cors pyngrok
from flask import Flask, request, jsonify, render_template_string
from flask_cors import CORS

app = Flask(__name__)
CORS(app)

SO = {   # cột số: (nhãn hiển thị, giá trị mặc định, bước nhảy)
    'Age': ('Tuổi', 28, 1),
    'FCVC': ('Tần suất ăn rau (1 đến 3)', 2, 1),
    'NCP': ('Số bữa chính mỗi ngày (1 đến 4)', 3, 1),
    'CH2O': ('Lượng nước uống (1 đến 3)', 2, 1),
    'FAF': ('Tần suất vận động (0 đến 3)', 1, 1),
    'TUE': ('Thời gian dùng thiết bị điện tử (0 đến 2)', 1, 1),
}
if DUNG_CHIEU_CAO_CAN_NANG:
    SO['Height'] = ('Chiều cao (m)', 1.70, 0.01)
    SO['Weight'] = ('Cân nặng (kg)', 70, 1)
CHON = {  # cột chữ: (nhãn hiển thị, các lựa chọn)
    'Gender': ('Giới tính', ['Male', 'Female']),
    'family_history_with_overweight': ('Gia đình có người thừa cân', ['yes', 'no']),
    'FAVC': ('Hay ăn đồ nhiều calo', ['yes', 'no']),
    'CAEC': ('Ăn vặt giữa các bữa', ['no', 'Sometimes', 'Frequently', 'Always']),
    'SMOKE': ('Hút thuốc', ['no', 'yes']),
    'SCC': ('Theo dõi lượng calo', ['no', 'yes']),
    'CALC': ('Uống rượu bia', ['no', 'Sometimes', 'Frequently']),
    'MTRANS': ('Phương tiện di chuyển', ['Public_Transportation', 'Walking', 'Automobile', 'Motorbike', 'Bike']),
}

def tao_form():
    h = ''
    for k, (nhan, mac_dinh, buoc) in SO.items():
        h += f'<label>{nhan}</label><input type="number" step="{buoc}" id="{k}" value="{mac_dinh}">'
    for k, (nhan, opts) in CHON.items():
        h += f'<label>{nhan}</label><select id="{k}">' + ''.join(f'<option>{o}</option>' for o in opts) + '</select>'
    return h

HTML_PAGE = """
<!DOCTYPE html>
<html lang="vi"><head><meta charset="UTF-8"><title>Dự đoán mức béo phì</title>
<style>
  body { font-family: Arial, sans-serif; max-width: 440px; margin: 30px auto; padding: 20px; }
  h2 { text-align: center; }
  label { display: block; margin-top: 10px; font-weight: bold; }
  input, select { width: 100%; padding: 6px; margin-top: 4px; box-sizing: border-box; }
  button { margin-top: 20px; width: 100%; padding: 10px; background: #2563eb; color: white; border: none; border-radius: 4px; cursor: pointer; }
  #result { margin-top: 20px; padding: 12px; border-radius: 4px; display: none; }
  .bar { display: flex; justify-content: space-between; font-size: 13px; margin-top: 4px; }
</style></head><body>
<h2>Dự đoán mức béo phì</h2>
__FORM__
<button onclick="predict()">Dự đoán</button>
<div id="result"></div>
<script>
const IDS = __IDS__;
async function predict() {
  const data = {};
  for (const id of IDS) { const el = document.getElementById(id); data[id] = el.type === 'number' ? parseFloat(el.value) : el.value; }
  const box = document.getElementById('result'); box.style.display = 'block'; box.style.background = '#f3f4f6'; box.textContent = 'Đang dự đoán...';
  try {
    const res = await fetch('/predict', {method: 'POST', headers: {'Content-Type': 'application/json'}, body: JSON.stringify(data)});
    const j = await res.json();
    if (!res.ok) throw new Error(j.error || 'Lỗi');
    box.style.background = '#e0f2fe';
    box.innerHTML = '<b>Dự đoán: ' + j.label + '</b>' + j.probs.map(([k, v]) => '<div class="bar"><span>' + k + '</span><span>' + v + '%</span></div>').join('');
  } catch (err) { box.style.background = '#fecaca'; box.textContent = 'Lỗi: ' + err.message; }
}
</script></body></html>
"""

@app.route('/')
def home():
    page = HTML_PAGE.replace('__FORM__', tao_form()).replace('__IDS__', str(list(SO) + list(CHON)))
    return render_template_string(page)

@app.route('/predict', methods=['POST'])
def predict():
    try:
        nhan, xs = du_doan(request.json)
        return jsonify({'label': nhan, 'probs': list(xs.items())})   # dạng danh sách để giữ đúng thứ tự các mức
    except Exception as e:
        return jsonify({'error': str(e)}), 400
```

```python
# Thử nhanh route /predict mà không cần mở ngrok
cl = app.test_client()
r = cl.post('/predict', json={'Gender': 'Female', 'Age': 22, 'family_history_with_overweight': 'no', 'FAVC': 'no',
                              'FCVC': 3, 'NCP': 3, 'CAEC': 'Sometimes', 'SMOKE': 'no', 'CH2O': 2, 'SCC': 'no',
                              'FAF': 2, 'TUE': 0, 'CALC': 'no', 'MTRANS': 'Walking', 'Height': 1.65, 'Weight': 55})
print(r.status_code, r.get_json())
```
