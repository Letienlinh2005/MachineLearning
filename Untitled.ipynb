```python
# Cell 10: Chuyển đổi dữ liệu
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X = df.iloc[:, :-1]
y = df.iloc[:, -1].map({n: i for i, n in enumerate(ORDER)}).astype(int)

X = X.drop(columns=['Height', 'Weight'])
X = pd.get_dummies(X, dtype=float)
feature_names = list(X.columns)
print("Số đặc trưng sau mã hoá:", len(feature_names))
X.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=42, stratify=y)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

joblib.dump(scaler, 'scaler.pkl')
joblib.dump(feature_names, 'feature_names.pkl')
print("Số lượng đặc trưng:", len(feature_names))

print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
pd.DataFrame(X_train, columns=feature_names).head()
```

```python
# Cell 11: Xây dựng và huấn luyện mạng nơ-ron
from tensorflow import keras
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.models import Sequential

model = Sequential([
    Input(shape=(X_train.shape[1],)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(len(ORDER), activation='softmax'),
])
model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()
```

```python
early_stop = keras.callbacks.EarlyStopping(
    monitor='val_loss', patience=10, restore_best_weights=True)

history = model.fit(X_train, y_train, epochs=100, batch_size=32,
                    validation_split=0.2, callbacks=[early_stop], verbose=1)

model.save('modelNN.h5')
```

```python
# Cell 12: Đồ thị loss và accuracy theo epoch
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(history.history['loss'], label='Train')
axes[0].plot(history.history['val_loss'], label='Validation')
axes[0].set_title("Loss theo epoch"); axes[0].set_xlabel("Epoch"); axes[0].legend()

axes[1].plot(history.history['accuracy'], label='Train')
axes[1].plot(history.history['val_accuracy'], label='Validation')
axes[1].set_title("Accuracy theo epoch"); axes[1].set_xlabel("Epoch"); axes[1].legend()
plt.tight_layout()
plt.savefig("hinh/loss_acc.png", dpi=150)
plt.show()
```

```python
# Cell 13: Đánh giá trên tập test (mục 3.5.3)
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

proba = model.predict(X_test, verbose=0)
pred = proba.argmax(axis=1)

print("Test Accuracy:", round(accuracy_score(y_test, pred), 4))
print(classification_report(y_test, pred, target_names=ORDER, digits=3))
```

```python
cm = confusion_matrix(y_test, pred)
plt.figure(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', linewidths=1, cbar=False,
            xticklabels=ORDER, yticklabels=ORDER)
plt.title("Ma trận nhầm lẫn (Confusion Matrix)", fontsize=15)
plt.xlabel("Dự đoán"); plt.ylabel("Thực tế")
plt.xticks(rotation=40, ha='right'); plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig("hinh/confusion_matrix.png", dpi=150)
plt.show()
```
